```python
print("")
print("*** Estatísticas da coluna CL_FHL: Número de filhos do cliente ***")
print("")
print("Contagem(count), média(mean), desvio padrão(std), mínimo(min), quartis(25% e 75%),  mediana(50%), máximo(max):")
print(varejo["CL_FHL"].describe())
print("")
print("*** Moda(mode) ***")
print(varejo["CL_FHL"].mode())
```

```python
Q1  = varejo["CL_FHL"].quantile(0.25)
Q3  = varejo["CL_FHL"].quantile(0.75)
IQR = Q3 - Q1

lim_inf = Q1 - 1.5 * IQR
lim_sup = Q3 + 1.5 * IQR

print(f"Faixa normal: [{lim_inf:.2f}, {lim_sup:.2f}]")

mascara = (varejo['CL_FHL'] < lim_inf) | (varejo['CL_FHL'] > lim_sup)
print("")
print(f"Outliers encontrados: {mascara.sum()}")
print("")
print(varejo[mascara][['CO_ID', 'CL_FHL']])

```

```python
# Calcular o Z-Score de uma coluna
media  = varejo['CL_FHL'].mean()    # média dos salários
desvio = varejo['CL_FHL'].std()     # desvio padrão

z_scores = (varejo['CL_FHL'] - media) / desvio
# z = 0   → valor igual à média
# z = 2   → 2 desvios ACIMA da média
# z = -3  → 3 desvios ABAIXO = outlier!

# Filtrar quem tem z muito alto ou muito baixo
outliers = varejo[z_scores.abs() > 3]
print("")
print(f"Outliers (|z| > 3): {len(outliers)}")
print("")
print(outliers[['CL_FHL']])
```
